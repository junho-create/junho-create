| ------------------------ | ---------------------------------------------------- | -------------------------------------------------------- | ---------------------------------------------------- |
| **Agent1 (Supervisor)**  | 사용자의 요청을 분석하고, 어떤 하위 Agent에게 보낼지 결정함 (라우팅 및 의도판단 담당) | `A1 RoutingAcc`, `A1 Coherence`                          | 올바른 에이전트로 분기하는 정확도, 판단 논리의 일관성                       |

1️⃣ A1 RoutingAcc (Routing Accuracy)

정의:
Supervisor가 사용자의 요청을 올바른 하위 에이전트(예: Agent2 예약, Agent3 증상추천 등)로 보냈는지의 비율.

기준:
실제 로그의 "output.routing_info.target" 값이, 미리 정의된 "expected" 매핑과 일치하는가.

평가 방식:

expected = {
    "reservation": "agent2_reservation",
    "symptom_doctor": "agent3_rag",
    "hospital_info": "tavily_search",
    "greeting": "agent1_direct",
}


→ 예를 들어 "primary_intent": "symptom_doctor" 이면
"target": "agent3_rag" 여야 정확함.

의미:
Supervisor의 의도 판단(Intention Classification) 능력,
즉 “이 사용자는 증상 문의 중이니 RAG agent로 보내야겠다” 같은 판단이 얼마나 정확한가를 측정함.

결과 예시 해석:

0.91 → 91%의 요청이 올바른 Agent로 라우팅됨.

0.5 → 반 정도는 엉뚱한 에이전트로 보냈다는 뜻.

2️⃣ A1 Coherence (Routing Coherence)

정의:
Supervisor의 의사결정 과정이 논리적이고 일관적인지 평가.

평가 방식:
OpenAI GPT가 다음과 같이 평가함:

prompt = f"Rate routing coherence (0~5).\nUser: {user_input}\nOutput: {agent_output}"


→ 즉, Supervisor가 내린 판단(예: “예약 관련 요청으로 분류”)이
사용자의 원래 입력과 논리적으로 맞는지를 점수로 매김.
(5점 만점 → 0~1 scale로 환산 시 5점은 1.0, 2.5점은 0.5)

의미:
Coherence = 논리 일관성 + 판단 근거의 자연스러움.
예를 들어 “어깨가 아파요” → “증상 관련 요청으로 분류”는 일관성 높음(5점).
반면 “예약 취소하고 싶어요” → “증상 관련 요청으로 분류”는 비일관성(0~1점).

🔹 Agent2 (Reservation) 평가 지표
3️⃣ A2 FormFill (Form Fill Accuracy)

정의:
Agent2가 사용자의 대화에서 예약에 필요한 필수 정보를 얼마나 잘 채웠는가 평가.

평가 방식:

REQ_FIELDS = ["환자명", "전화번호", "symptoms"]


각 로그의 output.collected_info 안에
이 3개 필드가 채워져 있으면 (예: {"환자명": "홍길동", "전화번호": "010...", "symptoms": "두통"})
→ 채운 개수 / 3으로 점수 계산.

의미:
Agent2의 정보 수집 능력, 즉 폼 자동 완성 정확도를 나타냄.

1.0 → 완벽히 모든 필드 채움.

0.33 → 필수 필드 중 하나만 채움.

4️⃣ A2 Parsing (Parsing Accuracy)

정의:
Agent2의 출력이 논리적 구조(JSON 형태) 로 제대로 파싱되었는지 여부.

평가 기준:
output 안에 "success": bool, "routing_info": dict 가 존재하면 성공으로 간주.
(즉, 내부 예약 처리 로직이 정상적으로 실행되었는가 확인)

의미:
Agent2의 예약 처리 코드 정상동작 여부 평가.
→ 실제 폼 데이터를 Supabase 등에 전달하기 전 단계까지 이상 없는지를 체크함.

🔹 Agent3 (RAG Doctor) 평가 지표
5️⃣ A3 Faithfulness (정확성 / 충실도)

정의:
Agent3의 답변이 RAG 검색 결과(컨텍스트)에 충실한가를 DeepEval이 판단.
(즉, hallucination이 없는가)

기준:
RAG retrieval context(증상-진료과 문서)와 비교해서
답변 내용이 근거 문서의 정보와 일치하면 1.0,
근거에 없는 내용을 말했으면 0.0.

의미:
Faithfulness = RAG 근거 충실도.
→ 예: “무릎이 아파요”일 때 정형외과 의사를 추천해야 하는데
“내과 전문의 추천”하면 Faithfulness 0점.

6️⃣ A3 Answer Relevancy (답변 관련성)

정의:
사용자의 질문과 Agent3의 답변이 주제적으로 얼마나 관련 있는가 평가.

기준:
DeepEval이 LLM을 사용해
"question" ↔ "answer" 간 의미적 유사도를 평가 (0~1).

의미:
Relevance = 사용자의 의도와의 일치도.
→ “어깨 아파요” → “정형외과 의사 추천” → 높은 관련성(1.0)
→ “어깨 아파요” → “예약은 오전 9시까지 가능합니다” → 낮은 관련성(0.0)

7️⃣ A3 Coherence (RAG Coherence)

정의:
Agent3의 응답 논리가 내부 reasoning 흐름과 일관되는지.
현재는 Supervisor coherence를 proxy로 사용 중 (sup_coh / 5).

의미:
향후 RAG reasoning trace 분석으로 대체 가능.
지금은 전반적 응답 일관성 지표 역할.

🔸 Overall Coh. (종합 일관성 지표)

정의:
시스템 전체가 사용자 → Supervisor → Reservation/RAG 순으로
얼마나 “논리적이고 자연스럽게” 협력했는가를 평가하는 종합 스코어.

계산식:

overall_coh = (Supervisor_Coherence + (Faithfulness + Relevance)/2) / 2


즉,

Supervisor의 판단 일관성

RAG 응답의 충실도 및 관련성
두 축을 평균내서 전체 에이전트 협업 품질을 측정.

🔹 Collaboration (협업/지연 분석)

communication_latency_mean_sec

로그 간 타임스탬프 차이의 평균값(초 단위)

Agent 간 응답 주기, 네트워크/LLM 지연의 평균 속도

communication_counts

총 메시지(턴)의 수 → Agent 간 대화 길이 / 복잡도 파악용

| **Agent2 (Reservation)** | 예약 요청을 처리함 (이름, 전화번호, 증상 등 폼 입력과 예약 파싱)              | `A2 FormFill`, `A2 Parsing`                              | 필요한 정보를 얼마나 잘 채웠는가, 예약 요청의 구조적 파싱 정확도                |
| **Agent3 (RAG Doctor)**  | 사용자의 증상에 맞는 의사·진료과를 추천 (RAG 기반 추론)                   | `A3 Faithfulness`, `A3 Answer Relevancy`, `A3 Coherence` | RAG 검색 결과에 얼마나 충실한가, 사용자 질문과 얼마나 관련된 답인가, 응답 논리의 일관성 |
| **전체 (Overall)**         | 에이전트 전체 체인의 협업 품질 평가                                 | `Overall Coh.`                                           | 각 단계(판단→행동→응답)의 평균적 품질 (coherence + relevance 기반)    |



1️⃣ A1 RoutingAcc (Routing Accuracy)

정의:
Supervisor가 사용자의 요청을 올바른 하위 에이전트(예: Agent2 예약, Agent3 증상추천 등)로 보냈는지의 비율.

기준:
실제 로그의 "output.routing_info.target" 값이, 미리 정의된 "expected" 매핑과 일치하는가.

평가 방식:

expected = {
    "reservation": "agent2_reservation",
    "symptom_doctor": "agent3_rag",
    "hospital_info": "tavily_search",
    "greeting": "agent1_direct",
}


→ 예를 들어 "primary_intent": "symptom_doctor" 이면
"target": "agent3_rag" 여야 정확함.

의미:
Supervisor의 의도 판단(Intention Classification) 능력,
즉 “이 사용자는 증상 문의 중이니 RAG agent로 보내야겠다” 같은 판단이 얼마나 정확한가를 측정함.

결과 예시 해석:

0.91 → 91%의 요청이 올바른 Agent로 라우팅됨.

0.5 → 반 정도는 엉뚱한 에이전트로 보냈다는 뜻.

2️⃣ A1 Coherence (Routing Coherence)

정의:
Supervisor의 의사결정 과정이 논리적이고 일관적인지 평가.

평가 방식:
OpenAI GPT가 다음과 같이 평가함:

prompt = f"Rate routing coherence (0~5).\nUser: {user_input}\nOutput: {agent_output}"


→ 즉, Supervisor가 내린 판단(예: “예약 관련 요청으로 분류”)이
사용자의 원래 입력과 논리적으로 맞는지를 점수로 매김.
(5점 만점 → 0~1 scale로 환산 시 5점은 1.0, 2.5점은 0.5)

의미:
Coherence = 논리 일관성 + 판단 근거의 자연스러움.
예를 들어 “어깨가 아파요” → “증상 관련 요청으로 분류”는 일관성 높음(5점).
반면 “예약 취소하고 싶어요” → “증상 관련 요청으로 분류”는 비일관성(0~1점).

🔹 Agent2 (Reservation) 평가 지표
3️⃣ A2 FormFill (Form Fill Accuracy)

정의:
Agent2가 사용자의 대화에서 예약에 필요한 필수 정보를 얼마나 잘 채웠는가 평가.

평가 방식:

REQ_FIELDS = ["환자명", "전화번호", "symptoms"]


각 로그의 output.collected_info 안에
이 3개 필드가 채워져 있으면 (예: {"환자명": "홍길동", "전화번호": "010...", "symptoms": "두통"})
→ 채운 개수 / 3으로 점수 계산.

의미:
Agent2의 정보 수집 능력, 즉 폼 자동 완성 정확도를 나타냄.

1.0 → 완벽히 모든 필드 채움.

0.33 → 필수 필드 중 하나만 채움.

4️⃣ A2 Parsing (Parsing Accuracy)

정의:
Agent2의 출력이 논리적 구조(JSON 형태) 로 제대로 파싱되었는지 여부.

평가 기준:
output 안에 "success": bool, "routing_info": dict 가 존재하면 성공으로 간주.
(즉, 내부 예약 처리 로직이 정상적으로 실행되었는가 확인)

의미:
Agent2의 예약 처리 코드 정상동작 여부 평가.
→ 실제 폼 데이터를 Supabase 등에 전달하기 전 단계까지 이상 없는지를 체크함.

🔹 Agent3 (RAG Doctor) 평가 지표
5️⃣ A3 Faithfulness (정확성 / 충실도)

정의:
Agent3의 답변이 RAG 검색 결과(컨텍스트)에 충실한가를 DeepEval이 판단.
(즉, hallucination이 없는가)

기준:
RAG retrieval context(증상-진료과 문서)와 비교해서
답변 내용이 근거 문서의 정보와 일치하면 1.0,
근거에 없는 내용을 말했으면 0.0.

의미:
Faithfulness = RAG 근거 충실도.
→ 예: “무릎이 아파요”일 때 정형외과 의사를 추천해야 하는데
“내과 전문의 추천”하면 Faithfulness 0점.

6️⃣ A3 Answer Relevancy (답변 관련성)

정의:
사용자의 질문과 Agent3의 답변이 주제적으로 얼마나 관련 있는가 평가.

기준:
DeepEval이 LLM을 사용해
"question" ↔ "answer" 간 의미적 유사도를 평가 (0~1).

의미:
Relevance = 사용자의 의도와의 일치도.
→ “어깨 아파요” → “정형외과 의사 추천” → 높은 관련성(1.0)
→ “어깨 아파요” → “예약은 오전 9시까지 가능합니다” → 낮은 관련성(0.0)

7️⃣ A3 Coherence (RAG Coherence)

정의:
Agent3의 응답 논리가 내부 reasoning 흐름과 일관되는지.
현재는 Supervisor coherence를 proxy로 사용 중 (sup_coh / 5).

의미:
향후 RAG reasoning trace 분석으로 대체 가능.
지금은 전반적 응답 일관성 지표 역할.

🔸 Overall Coh. (종합 일관성 지표)

정의:
시스템 전체가 사용자 → Supervisor → Reservation/RAG 순으로
얼마나 “논리적이고 자연스럽게” 협력했는가를 평가하는 종합 스코어.

계산식:

overall_coh = (Supervisor_Coherence + (Faithfulness + Relevance)/2) / 2


즉,

Supervisor의 판단 일관성

RAG 응답의 충실도 및 관련성
두 축을 평균내서 전체 에이전트 협업 품질을 측정.

🔹 Collaboration (협업/지연 분석)

communication_latency_mean_sec

로그 간 타임스탬프 차이의 평균값(초 단위)

Agent 간 응답 주기, 네트워크/LLM 지연의 평균 속도

communication_counts

총 메시지(턴)의 수 → Agent 간 대화 길이 / 복잡도 파악용

| Metric                  | Agent       | 평가 대상     | 판단 기준                                    |
| :---------------------- | :---------- | :-------- | :--------------------------------------- |
| **A1 RoutingAcc**       | Supervisor  | 의도 분류 정확도 | 기대 target과 실제 routing 일치율                |
| **A1 Coherence**        | Supervisor  | 판단 논리 일관성 | 사용자 입력 대비 응답의 논리 자연스러움                   |
| **A2 FormFill**         | Reservation | 정보 수집 정확도 | 필수 폼 필드(3개) 채움 비율                        |

| **A3 Faithfulness**     | RAG         | 사실 충실도    | retrieval context 기반 일치율                 |
| **A3 Answer Relevancy** | RAG         | 질문-답변 관련성 | question-answer 의미 유사도                   |

| **Overall Coh.**        | 전체          | 협업 품질     | (Supervisor Coherence + (Faith+Rel)/2)/2 |


```python
import matplotlib.pyplot as plt

# ---------------------------
# 데이터
# ---------------------------
metrics = {
    "A1 RoutingAcc": 0.9167,
    "A1 Coherence": 0.96,
    "A2 FormFill": 0.2837,
    "A2 Parsing": 1.0,
    "A3 Faithfulness (DeepEval)": 0.5455,
    "A3 Relevancy (RAGAS)": 0.7228,
    "Overall Coh. (DeepEval)": 0.8437
}

samples = {
    "A1 RoutingAcc": 96,
    "A1 Coherence": 10,
    "A2 FormFill": 47,
    "A2 Parsing": 47,
    "A3 Faithfulness (DeepEval)": 11,
    "A3 Relevancy (RAGAS)": 63,
    "Overall Coh. (DeepEval)": 21
}

# 각 metric별 사용 라이브러리 구분
methods = {
    "A1 RoutingAcc": "Custom",
    "A1 Coherence": "LLM-as-a-Judge",
    "A2 FormFill": "Custom",
    "A2 Parsing": "Custom",
    "A3 Faithfulness (DeepEval)": "DeepEval",
    "A3 Relevancy (RAGAS)": "RAGAS",
    "Overall Coh. (DeepEval)": "DeepEval"
}

# ---------------------------
# 색상 팔레트 (기법별 고정)
# ---------------------------
colors = {
    "Custom": "#6BAED6",          # 파란색
    "LLM-as-a-Judge": "#74C476",  # 초록색
    "RAGAS": "#9E9AC8",           # 보라색
    "DeepEval": "#FB6A4A"         # 빨강색
}

# ---------------------------
# 시각화 설정
# ---------------------------
plt.rcParams["font.family"] = "Malgun Gothic"
plt.rcParams["axes.unicode_minus"] = False

fig, ax = plt.subplots(figsize=(10, 6))
bar_colors = [colors[methods[m]] for m in metrics.keys()]

bars = ax.bar(metrics.keys(), metrics.values(),
              color=bar_colors, edgecolor="black", linewidth=0.8)

# ---------------------------
# 그래프 스타일링
# ---------------------------
ax.set_title("Agent Evaluation Summary (기법별 구분)", fontsize=16, fontweight="bold", pad=25)
ax.set_ylabel("Score (0 ~ 1)", fontsize=12)
ax.set_ylim(0, 1.05)
ax.set_xticks(range(len(metrics)))
ax.set_xticklabels(metrics.keys(), rotation=20, ha="right", fontsize=10)
ax.grid(axis="y", linestyle="--", alpha=0.4)

# 막대 위에 점수 및 샘플 수 표시
for bar, label, value in zip(bars, metrics.keys(), metrics.values()):
    n = samples[label]
    ax.text(
        bar.get_x() + bar.get_width()/2,
        value + 0.02,
        f"{value:.2f}\n(n={n})",
        ha="center", va="bottom", fontsize=9, fontweight="semibold"
    )

# ---------------------------
# 범례 추가 (기법별 색상)
# ---------------------------
handles = [
    plt.Rectangle((0, 0), 1, 1, color=colors[name], label=name)
    for name in colors
]
ax.legend(
    handles=handles,
    title="평가 기법",
    fontsize=9,
    title_fontsize=10,
    loc="lower right",
    frameon=True
)

# ---------------------------
# 미세 조정
# ---------------------------
for spine in ["top", "right"]:
    ax.spines[spine].set_visible(False)
ax.set_facecolor("#FAFAFA")

plt.tight_layout()
plt.show()
```
